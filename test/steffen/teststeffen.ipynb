```python
import pandas as pd
from pathlib import Path # Importiere Path für die Pfadmanipulation

def get_unique_entities(file_path):
    """
    Liest eine CSV-Datei, extrahiert die 'Entity'-Spalte
    und gibt eine sortierte Liste der einzigartigen Einträge zurück.
    """
    try:
        # Sicherstellen, dass der Pfad korrekt ist und existiert
        if not Path(file_path).exists():
            print(f"Fehler: Die Datei '{file_path}' wurde nicht gefunden.")
            return []

        df = pd.read_csv(file_path)

        if 'Entity' in df.columns:
            unique_entities = df['Entity'].unique().tolist()
            unique_entities.sort() # Sortiere die Liste alphabetisch für bessere Lesbarkeit
            return unique_entities
        else:
            print("Fehler: Die Spalte 'Entity' wurde in der CSV-Datei nicht gefunden.")
            return []
    except Exception as e:
        print(f"Ein Fehler ist aufgetreten beim Lesen der Datei oder der Spalte: {e}")
        return []

if __name__ == "__main__":
    # Verwende den von dir angegebenen absoluten Pfad direkt
    csv_file_path = r"C:\Users\Admin\Desktop\Vorlesungen\1.11 Datenvisualisierung mit Python\20250714 Projekt\Projekt-Emission-Dashboard\data\merged_final.csv"

    print(f"Versuche, die Datei zu laden von: {csv_file_path}")

    unique_list = get_unique_entities(csv_file_path)

    if unique_list:
        print("\nEinzigartige Einträge in der 'Entity'-Spalte:")
        for entity in unique_list:
            print(entity)
```

```python
import pandas as pd
from pathlib import Path

def extract_and_save_unique_entities(input_csv_path, output_csv_path):
    """
    Liest eine CSV-Datei, extrahiert die 'Entity'-Spalte,
    findet einzigartige Einträge und speichert sie in einer neuen CSV-Datei.
    """
    try:
        # Sicherstellen, dass der Pfad der Eingabedatei korrekt ist und existiert
        input_path_obj = Path(input_csv_path)
        if not input_path_obj.exists():
            print(f"Fehler: Die Eingabedatei '{input_csv_path}' wurde nicht gefunden.")
            return False

        # pandas ist intelligent genug, um mit Path-Objekten umzugehen
        df = pd.read_csv(input_csv_path)

        if 'Entity' in df.columns:
            unique_entities = df['Entity'].unique().tolist()
            unique_entities.sort() # Sortiere die Liste alphabetisch

            # Erstelle ein neues DataFrame nur mit den einzigartigen Entitäten
            unique_df = pd.DataFrame({'Entity': unique_entities})

            # Speichere das neue DataFrame in einer CSV-Datei
            # output_csv_path ist hier bereits ein Path-Objekt oder ein String, der funktioniert
            unique_df.to_csv(output_csv_path, index=False, encoding='utf-8')
            print(f"Einzigartige Entitäten wurden erfolgreich gespeichert in: {output_csv_path}")
            return True
        else:
            print("Fehler: Die Spalte 'Entity' wurde in der CSV-Datei nicht gefunden.")
            return False
    except Exception as e:
        print(f"Ein Fehler ist aufgetreten: {e}")
        return False

if __name__ == "__main__":
    # Pfad zur Eingabe-CSV-Datei (deine ursprüngliche merged_final.csv)
    input_file = r"C:\Users\Admin\Desktop\Vorlesungen\1.11 Datenvisualisierung mit Python\20250714 Projekt\Projekt-Emission-Dashboard\data\merged_final.csv"

    # Pfad für die Ausgabedatei (die neue CSV mit den einzigartigen Entitäten)
    # Diese Datei wird im aktuellen Arbeitsverzeichnis deines Notebooks/Python-Interpreters gespeichert.
    # Du kannst den Namen auch in 'laender.csv' ändern, wenn du möchtest.
    output_file = "unique_entities.csv" # Oder Path("unique_entities.csv") wenn du explizit Path-Objekte bevorzugst

    print(f"Versuche, die Original-CSV-Datei zu laden von: {input_file}")

    success = extract_and_save_unique_entities(input_file, output_file)

    if success:
        print(f"\nBitte öffne die Datei '{output_file}' (sie sollte im selben Ordner sein, wo du dieses Skript ausführst) und kopiere den gesamten Inhalt hier in den Chat.")
    else:
        print("\nEs gab ein Problem beim Extrahieren oder Speichern der Entitäten. Bitte überprüfe die Fehlermeldungen oben.")
```
